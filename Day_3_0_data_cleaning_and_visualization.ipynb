```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("heart_disease_risk_dataset_earlymed.csv")
print(df.shape)
print(df.head())
print(df.describe)
```

```python
print(df.isnull().sum())
```

```python
print('Duplicactes', df.duplicated().sum())
```

```python
# Read the dataset from Excel format
df = pd.read_excel("heart_disease_risk_dataset_earlymed.xls")
print(df.head())
```

```python
print("Duplicates", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

```python
print(df.isnull().sum())
```

```python
import seaborn as sns
plt.figure()
sns.displot(df["age"], kde=True)
plt.title("Age across all genders")
plt.show()
```

```python
for col in ['obesity', 'cholesterol_high', 'palpitations', 'family_history']:
  plt.figure()
  sns.displot(x=df[col])
  plt.show()
```

```python
# Set the figure size for the heatmap plot
plt.figure(figsize=(10, 6))

# Calculate the Pearson correlation matrix for only the numeric columns in the DataFrame
corr = df.corr(numeric_only=True)

# Generate a heatmap using Seaborn, annotating each cell with its correlation coefficient value
sns.heatmap(corr, annot=True)

# Display the plotted heatmap
plt.show()
```
